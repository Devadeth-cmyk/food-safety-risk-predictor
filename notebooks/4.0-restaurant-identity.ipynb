```python
df["Address_normalized"] = (
    df["Address_clean"]
    .astype(str)
    .str.upper()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("Original:")
print(df["Address"].head(10).tolist())

print("\nNormalized:")
print(df["Address_normalized"].head(10).tolist())
```

```python
id_address_normalized_counts = (
    df.groupby("Restaurant_ID")["Address_normalized"]
      .nunique()
)

print("Restaurant_IDs with exactly 1 normalized address:",
      (id_address_normalized_counts == 1).sum())

print("Restaurant_IDs with multiple normalized addresses:",
      (id_address_normalized_counts > 1).sum())

print("Maximum normalized addresses for one Restaurant_ID:",
      id_address_normalized_counts.max())
```

```python
multi_address_ids_normalized = (
    id_address_normalized_counts[
        id_address_normalized_counts > 1
    ].index
)

inspection_view = (
    df[df["Restaurant_ID"].isin(multi_address_ids_normalized[:15])]
    [
        [
            "Restaurant_ID",
            "Address",
            "Address_normalized",
            "Inspection Date",
            "Results"
        ]
    ]
    .sort_values(
        ["Restaurant_ID", "Inspection Date"]
    )
)

print(inspection_view.to_string(index=False))
```

```python
# Get the remaining Restaurant_IDs that have multiple normalized addresses
multi_address_ids_normalized = (
    id_address_normalized_counts[
        id_address_normalized_counts > 1
    ].index
)

# Get unique address variants for those IDs
address_variants = (
    df[df["Restaurant_ID"].isin(multi_address_ids_normalized)]
    [["Restaurant_ID", "Address_normalized"]]
    .drop_duplicates()
    .sort_values(["Restaurant_ID", "Address_normalized"])
)

print("Number of unique Restaurant_ID + address combinations:",
      len(address_variants))

print("\nFirst 100 address variants:")
print(address_variants.head(100).to_string(index=False))
```

```python
# Find address pairs where one address is exactly the other
# with an additional common street-type suffix.

street_types = {
    "ST", "AVE", "RD", "DR", "BLVD",
    "PL", "PKWY", "PLZ", "HWY", "CT"
}

suffix_variants = []

for restaurant_id, group in (
    address_variants.groupby("Restaurant_ID")
):
    addresses = group["Address_normalized"].tolist()

    for address in addresses:
        parts = address.split()

        if not parts:
            continue

        last_token = parts[-1]

        if last_token in street_types:
            shorter = " ".join(parts[:-1])

            if shorter in addresses:
                suffix_variants.append(
                    {
                        "Restaurant_ID": restaurant_id,
                        "Shorter_address": shorter,
                        "Full_address": address,
                        "Suffix": last_token
                    }
                )

suffix_variants_df = pd.DataFrame(suffix_variants)

print("Potential omitted street-suffix cases:",
      len(suffix_variants_df))

print("\nCases found:")
print(suffix_variants_df.to_string(index=False))
```

```python
inspect_ids = [1975.0, 1909538.0, 2192579.0]

suffix_inspection_history = (
    df[
        df["Restaurant_ID"].isin(inspect_ids)
    ][
        [
            "Restaurant_ID",
            "Address",
            "Address_normalized",
            "Inspection Date",
            "Inspection Type",
            "Results"
        ]
    ]
    .sort_values(
        ["Restaurant_ID", "Inspection Date"]
    )
)

print(
    suffix_inspection_history.to_string(index=False)
)
```

```python
from difflib import SequenceMatcher

address_pairs = []

for restaurant_id, group in address_variants.groupby("Restaurant_ID"):
    addresses = group["Address_normalized"].tolist()

    if len(addresses) < 2:
        continue

    for i in range(len(addresses)):
        for j in range(i + 1, len(addresses)):
            a = addresses[i]
            b = addresses[j]

            similarity = SequenceMatcher(
                None, a, b
            ).ratio()

            address_pairs.append({
                "Restaurant_ID": restaurant_id,
                "Address_1": a,
                "Address_2": b,
                "Similarity": similarity
            })

address_similarity_df = pd.DataFrame(address_pairs)

address_similarity_df = (
    address_similarity_df
    .sort_values(
        ["Similarity", "Restaurant_ID"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

print("Address pairs:", len(address_similarity_df))

print(
    address_similarity_df[
        address_similarity_df["Similarity"] >= 0.80
    ].to_string(index=False)
)
```
